# Travel Tax: NBA Schedule Friction & Competitive Equity

Five-season geospatial study of 6,000 NBA regular-season games.

## Research question
Does relative pre-game schedule burden predict game performance after accounting for recent team form, season, and team effects?

## Schedule Friction Index
The SFI combines travel distance, time-zone movement, rest deficit, seven-day game density, road streak, and altitude gain. All components are measured before the game and standardized within season.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path("..")
findings = json.loads((ROOT/"outputs"/"findings.json").read_text())
road = pd.read_csv(ROOT/"outputs"/"road_deciles.csv")
teams = pd.read_csv(ROOT/"outputs"/"team_burden.csv")
findings
```

## Main result
The verified fixed-effects model estimates **-1.35 points** per +1 unit of relative SFI (95% CI -2.06 to -0.63; p=0.0002348).

The average modeled change in win probability is **-4.0 percentage points**.
